```python
import numpy as np
import matplotlib.pyplot as plt

# Load the binary data (374,400 float32 values)
data = np.fromfile("input_1.bin", dtype=np.float32)

# Reshape to expected dimensions [1][3][520][240]
data = data.reshape(1, 3, 520, 240)

# If this is image data, visualize it
plt.figure(figsize=(10, 8))
# Display as RGB image (transpose to get proper dimensions)
rgb_image = np.transpose(data[0], (1, 2, 0))
# Normalize if needed
print(rgb_image.max())
if rgb_image.max() > 1.0:
    rgb_image = rgb_image / 255.0
plt.imshow(rgb_image)



plt.savefig("visualization.png")
plt.show()
```
